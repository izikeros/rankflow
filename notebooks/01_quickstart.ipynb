# 01 -- Quick Start

This notebook covers the basics: creating a `RankFlow` object from raw data or a pandas DataFrame, rendering a rank evolution plot, and customizing the visual appearance via `PlotConfig`.

**Prerequisites:** `pip install rankflow` (matplotlib and numpy are included).

```python
%matplotlib inline

import numpy as np
import pandas as pd

from rankflow import RankFlow, PlotConfig
```

## Creating RankFlow from a numpy array

The simplest input is a 2-D array of shape `(n_steps, n_documents)` where each value is the rank of that document at that step. Lower rank = better position.

```python
ranks = np.array([
    [0, 1, 2, 3, 4],  # Step 0: initial retrieval
    [2, 0, 1, 4, 3],  # Step 1: after re-ranking
    [1, 2, 0, 3, 4],  # Step 2: after second re-ranking
])

rf = RankFlow(ranks=ranks)
rf.plot()
```

## Adding custom labels

By default, steps are labeled "Step 0", "Step 1", ... and chunks are "Chunk 0", "Chunk 1", .... You can supply your own labels.

```python
rf = RankFlow(
    ranks=ranks,
    step_labels=["BM25", "Semantic", "Cross-Encoder"],
    chunk_labels=["doc_a", "doc_b", "doc_c", "doc_d", "doc_e"],
)
rf.plot()
```

## Attaching pipeline configuration

You can attach a `pipeline_config` dict describing the settings that produced this ranking. This metadata is preserved through JSON export/import and is used for automatic config diffing when comparing experiments (see [notebook 06](06_experiments_and_comparison.ipynb)).

```python
rf = RankFlow(
    ranks=ranks,
    step_labels=["BM25", "Semantic", "Cross-Encoder"],
    chunk_labels=["doc_a", "doc_b", "doc_c", "doc_d", "doc_e"],
    pipeline_config={"retriever": "bm25", "top_k": 100, "chunk_size": 512},
)
print(f"Config: {rf.pipeline_config}")
```

## Creating RankFlow from a pandas DataFrame

If your data is already in a DataFrame, pass it directly. Row index becomes step labels, column names become chunk labels.

```python
df = pd.DataFrame(
    ranks,
    index=["BM25", "Semantic", "Cross-Encoder"],
    columns=["doc_a", "doc_b", "doc_c", "doc_d", "doc_e"],
)
df
```

```python
rf = RankFlow(df=df)
rf.plot()
```

## Customizing the plot with PlotConfig

`PlotConfig` is a dataclass holding all visual settings. You can pass any of its fields as keyword arguments to `RankFlow()`.

```python
rf = RankFlow(
    df=df,
    title="My RAG Pipeline",
    title_font_size=16,
    caption="Retrieval stage",
    line_width=25,
    colors=["#e63946", "#457b9d", "#2a9d8f", "#e9c46a", "#f4a261"],
)
rf.plot()
```

You can also create a `PlotConfig` directly and reuse it across multiple plots.

```python
cfg = PlotConfig(
    title="Shared Config",
    fig_size=(6, 4),
    line_width=15,
    chunk_label_font_size=9,
)

# Use the config on different data
rf1 = RankFlow(df=df, **cfg.to_dict())
rf1.plot()
```
